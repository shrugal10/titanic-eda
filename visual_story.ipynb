```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data/Titanic-Dataset.csv")

sns.set_theme(style="whitegrid")

df.head()
```

# Titanic Visual Deep-Dive

A publication-quality visual exploration of the factors associated with Titanic passenger survival.

```python


plt.figure(figsize=(8, 5))

ax = sns.countplot(data=df, x="Survived")

ax.set_title("Titanic Passenger Survival Overview", fontsize=16, fontweight="bold")
ax.set_xlabel("Survival Status")
ax.set_ylabel("Number of Passengers")

ax.set_xticks([0, 1])
ax.set_xticklabels(["Did Not Survive", "Survived"])

plt.tight_layout()
plt.show()
```

**Takeaway:** Most passengers aboard the Titanic did not survive the disaster.

```python


plt.figure(figsize=(8, 5))

ax = sns.countplot(data=df, x="Sex", hue="Survived")

ax.set_title("Titanic Survival by Sex", fontsize=16, fontweight="bold")
ax.set_xlabel("Sex")
ax.set_ylabel("Number of Passengers")

ax.legend(title="Survival Status", labels=["Did Not Survive", "Survived"])

plt.tight_layout()
plt.show()
```

**Takeaway:** Female passengers had a much higher survival count relative to their total number than male passengers.

```python
plt.figure(figsize=(8, 5))

ax = sns.countplot(data=df, x="Pclass", hue="Survived")

ax.set_title("Titanic Survival by Passenger Class", fontsize=16, fontweight="bold")
ax.set_xlabel("Passenger Class")
ax.set_ylabel("Number of Passengers")

ax.legend(title="Survival Status", labels=["Did Not Survive", "Survived"])

plt.tight_layout()
plt.show()
```

**Takeaway:** Passengers in first class had a much higher chance of survival than passengers in lower classes.

```python
plt.figure(figsize=(8, 5))

ax = sns.histplot(data=df, x="Age", hue="Survived", bins=30, kde=True)

ax.set_title("Age Distribution of Titanic Passengers by Survival", fontsize=16, fontweight="bold")
ax.set_xlabel("Age")
ax.set_ylabel("Number of Passengers")

ax.legend(title="Survival Status", labels=["Survived", "Did Not Survive"])

plt.tight_layout()
plt.show()
```

**Takeaway:** Passengers who survived and those who did not show overlapping age distributions, although younger passengers were also present among the survivors.

```python
plt.figure(figsize=(8, 5))

ax = sns.boxplot(data=df, x="Survived", y="Fare")

ax.set_title("Fare Distribution by Titanic Survival", fontsize=16, fontweight="bold")
ax.set_xlabel("Survival Status")
ax.set_ylabel("Fare")

ax.set_xticks([0, 1])
ax.set_xticklabels(["Did Not Survive", "Survived"])

plt.tight_layout()
plt.show()
```

**Takeaway:** Survivors generally paid higher fares than passengers who did not survive, suggesting that fare and passenger class were related to survival.

```python
plt.figure(figsize=(9, 5))

ax = sns.barplot(data=df, x="Pclass", y="Survived", hue="Sex")

ax.set_title("Titanic Survival Rate by Passenger Class and Sex", fontsize=16, fontweight="bold")
ax.set_xlabel("Passenger Class")
ax.set_ylabel("Survival Rate")

ax.set_ylim(0, 1)

plt.tight_layout()
plt.show()
```

**Takeaway:** Survival rates were highest among females in first class and lowest among males in third class, showing the combined effect of sex and passenger class.

## Overall Survival Story

The Titanic survival data shows that survival was strongly associated with both passenger sex and social class. Overall, more passengers died than survived, showing how severe the disaster was. The survival-by-sex chart reveals a major difference between male and female passengers: females had substantially better survival outcomes, while most male passengers did not survive. Passenger class also had a clear relationship with survival. First-class passengers had the strongest outcomes, while third-class passengers experienced the lowest survival counts and rates. The age distribution shows that survivors and non-survivors came from overlapping age groups, although children and younger passengers were represented among those who survived. Fare provides another useful perspective, with survivors generally having higher fares and a wider fare distribution than passengers who did not survive. This likely reflects the relationship between fare and passenger class. Finally, comparing sex and class together shows the strongest contrast: females in first and second class had very high survival rates, while males, particularly those in third class, had much lower rates. Overall, the visual evidence suggests that survival was not random; sex, passenger class, and related socioeconomic factors played an important role in determining who survived the Titanic disaster.
